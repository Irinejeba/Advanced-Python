### **Python Numpy Array**

```python
import numpy as np
```

```python
array_1 =np.array([10,20,30,40])
print(array_1)
print (array_1.shape)
print (array_1.size)
print (array_1.ndim)
print (array_1.dtype)
```

```python
array_2 = np.array([[10,20,30],[40,50,60],[70,80,90]])
print(array_2)
print (array_2.shape)
print (array_2.size)
print (array_2.ndim)
print (array_2.dtype)
```

```python
array_3 = np.array([[[10,20,30],[40,50,60]],[[10,20,30],[40,50,60]]])
print(array_3)
print (array_3.shape)
print (array_3.size)
print (array_3.ndim)
print (array_3.dtype)
```

### **Type Casting**

```python
array_2 = np.array([[10,20,30],[40,50,60],[70,"Hai",90]],dtype="int")
print (array_2)
print (array_2.dtype)
```

```python
array_2 = np.array([[10,20,30],[40,50,60],[70,"100",90]],dtype="int")
print (array_2)
print (array_2.dtype)
```

```python
array_2 = np.array([[10,20,30],[40,50,60],[70,"100",90]],dtype="float")
print (array_2)
print (array_2.dtype)
```

### **Array Filling**

```python
a= np.full((5,5),2)
print (a)
```

```python
a= np.full((2,5,5),2)
print (a)
```

```python
b= np.zeros((5,5),dtype='int')
print (b)
```

```python
c=np.ones((5,5))
print (c)
```
