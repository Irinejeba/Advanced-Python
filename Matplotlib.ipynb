### **Python Matplotlib**

```python
import matplotlib.pyplot as plt
```

```python
plt.figure()
plt.plot()
plt.show();
```

```python
import numpy as np
```

### **Scatter Plat**

```python
x= np.array([1,2,3,4])
y= np.array([1,2,3,4])
plt.scatter(x,y,c='r',marker ='*',s= 100)
plt.show();
```

```python
x= np.array([1,2,3,4])
y= np.array([1,2,3,4])
color =['blue','yellow','red','green']
plt.scatter(x,y,c=color,marker ='*',s= 100)
plt.show();
```

### **Line graph**

```python
x= np.array([1,2,3,4])
y= np.array([1,2,3,4])
color =['blue','yellow','red','green']
plt.plot(x,y, c='g', linestyle ='dashed')
plt.scatter(x,y,c=color,marker="*",s=100)
plt.show();
```

### **Bar Chart**

```python
year =["1990","2000","2010","2020","2030"]
lang =["None","Java","C","C++","Python"]
c= ["Green","darkorange","violet","red","blue"]
plt.bar(year,lang,color= c,align="edge")
plt.show();
```

```python
#frequency distribution
np.random.seed(2)
a = np.random.normal(80,1.5,100)
plt.hist(a, color='deeppink',bins=15,edgecolor="Black")
plt.show();
```

### **Pie Chart**

```python
lang = ["Python","Java","C++","C"]
Votes = [90,75,50,40]
exp = [0.1,0,0.2,0]
c= ['Green','darkorange','deeppink','violet']
plt.pie(Votes,labels =lang,explode= exp,colors=c)
plt.show();
```

```python
years =["2000","2010","2020","2030"]
income = [200,400,600,800]
plt.title("Daily Income of Irine")
plt.xlabel("Years")
plt.xlabel("Income")
plt.plot(years,income,c='g',marker="*")
plt.show();
```
