### **Scatter Plat**

```python
x= np.array([1,2,3,4])
y= np.array([1,2,3,4])
plt.scatter(x,y,c='r',marker ='*',s= 100)
plt.show();
```

```python
x= np.array([1,2,3,4])
y= np.array([1,2,3,4])
color =['blue','yellow','red','green']
plt.scatter(x,y,c=color,marker ='*',s= 100)
plt.show();
```

### **Line graph**

```python
x= np.array([1,2,3,4])
y= np.array([1,2,3,4])
color =['blue','yellow','red','green']
plt.plot(x,y, c='g', linestyle ='dashed')
plt.scatter(x,y,c=color,marker="*",s=100)
plt.show();
```

### **Bar Chart**

```python
year =["1990","2000","2010","2020","2030"]
lang =["None","Java","C","C++","Python"]
c= ["Green","darkorange","violet","red","blue"]
plt.bar(year,lang,color= c,align="edge")
plt.show();
```

```python
#frequency distribution
np.random.seed(2)
a = np.random.normal(80,1.5,100)
plt.hist(a, color='deeppink',bins=15,edgecolor="Black")
plt.show();
```

### **Pie Chart**

```python
lang = ["Python","Java","C++","C"]
Votes = [90,75,50,40]
exp = [0.1,0,0.2,0]
c= ['Green','darkorange','deeppink','violet']
plt.pie(Votes,labels =lang,explode= exp,colors=c)
plt.show();
```

```python
years =["2000","2010","2020","2030"]
income = [200,400,600,800]
plt.title("Daily Income of Irine")
plt.xlabel("Years")
plt.xlabel("Income")
plt.plot(years,income,c='g',marker="*")
plt.show();
```

```python
#Stock Market
StockA =[1,2,3,4,5]
StockB =[3,4,5,3,2]
StockC =[6,4,2,1]
plt.title("Stock Market")
plt.plot(StockA,label="Stock 1",marker="*")
plt.plot(StockB,label="Stock 2",marker="p")
plt.plot(StockC,label="Stock 3",marker="o")
plt.legend()
plt.show();
```

### **Subplots comparing (age vs chol),(age vs thalach)**

```python
fig,ax=plt.subplots(nrows=2,ncols=1,figsize=(10,6))
#subplot=1
scatter =ax[0].scatter(df["age"],df["chol"],c=df["target"],cmap="winter")
ax[0].set(title='Age vs Cholestrol',xlabel="Age",ylabel="Thalach")
ax[0].legend(*scatter.legend_elements(),title="Target")
#add mean line
ax[0].axhline(df["chol"].mean(),color='r',linestyle="dashed")
    
scatter =ax[1].scatter(df["age"],df["thalach"],c=df["target"],cmap="viridis")
#customise the plot
ax[1].set(title='Age vs Thalch',xlabel="Age",ylabel="Thalach")
ax[1].legend(*scatter.legend_elements(),title="Target")
ax[1].axhline(df["thalach"].mean(),color="r",linestyle="dashed")
plt.subplots_adjust(hspace=0.4)
plt.show();
```

```python
fig,ax=plt.subplots(nrows=1,ncols=1,figsize=(10,6))
#subplot=1
scatter =ax.scatter(df["age"],df["chol"],c=df["target"],cmap="winter")
ax.set(title='Heart Disease and Cholestrol Levels',xlabel="Age",ylabel="Cholestrol")
ax.legend(*scatter.legend_elements(),title="Target")
#add mean line
ax.axhline(df["chol"].mean(),linestyle="dashed");
plt.savefig("Probability_of_heart_attack.png")
```
